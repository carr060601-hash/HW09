```python
# HW09 兩場活動報名整理器
# 姓名：蕭雅薰
# 學號：D1382798

# 固定活動名稱（tuple）
activity_names = ("桌遊體驗", "電影欣賞")

# 原始報名資料
board_raw = " Amy, Ben, amy, Cora "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 三組測試資料
tests = [
    ("測試一", " ben, Dora, ERIC, ben "),
    ("測試二", " ben, Dora, ERIC, ben, Frank "),
    ("測試三", " AMY, ben, CORA, amy ")
]

# 依序執行三組測試
for test_name, movie_raw in tests:

    print(f"\n===== {test_name} =====")

    # 1. 清除空白並統一小寫（list）
    board_list = board_raw.replace(" ", "").lower().split(",")
    movie_list = movie_raw.replace(" ", "").lower().split(",")

    # 2. 使用 set 去除重複
    board_set = set(board_list)
    movie_set = set(movie_list)

    # 3. 集合運算
    common = board_set & movie_set
    all_people = board_set | movie_set
    only_board = board_set - movie_set
    only_movie = movie_set - board_set

    # 4. 排序
    board_unique = sorted(board_set)
    movie_unique = sorted(movie_set)
    common_list = sorted(common)
    all_list = sorted(all_people)
    only_board_list = sorted(only_board)
    only_movie_list = sorted(only_movie)
    preview = all_list[:2]

    # 5. 使用 dict 儲存統計結果
    summary = {
        "桌遊人數": len(board_set),
        "電影人數": len(movie_set),
        "共同人數": len(common),
        "總不重複人數": len(all_people),
        "報名人次": len(board_set) + len(movie_set),
        "桌遊應收": len(board_set) * unit_fees[activity_names[0]],
        "電影應收": len(movie_set) * unit_fees[activity_names[1]]
    }

    summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

    # 6. 輸出結果
    print("HW09 活動報名摘要")
    print(f"活動：{activity_names[0]}，每人 {unit_fees[activity_names[0]]:.2f} 元")
    print(f"活動：{activity_names[1]}，每人 {unit_fees[activity_names[1]]:.2f} 元")
    print("桌遊整理串列：", board_list)
    print("電影整理串列：", movie_list)
    print("桌遊去重名單：", board_unique)
    print("電影去重名單：", movie_unique)
    print("桌遊人數：", summary["桌遊人數"], "人")
    print("電影人數：", summary["電影人數"], "人")
    print("共同參加者：", common_list)
    print("共同人數：", summary["共同人數"], "人")
    print("只參加桌遊者：", only_board_list)
    print("只參加電影者：", only_movie_list)
    print("全部參加者：", all_list)
    print("總不重複人數：", summary["總不重複人數"], "人")
    print("報名人次：", summary["報名人次"], "人次")
    print("通知名單前兩位：", preview)
    print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
    print(f"電影應收：{summary['電影應收']:.2f} 元")
    print(f"總應收：{summary['總應收']:.2f} 元")
```
